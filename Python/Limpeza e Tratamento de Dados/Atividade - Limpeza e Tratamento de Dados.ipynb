Tratar dados 'tempo.csv'

Aparencia: Sol, Nublado e chuva |
Temperatura: -130 ~ 130F |
Umidade: 0 ~ 100 | 
Jogar: Sim ou Nao

Importar dados

```python
import pandas as pd
import statistics as sts
```

```python
dataset = pd.read_csv('data/tempo.csv', sep=';')
dataset.head()
```

Verificar coluna Aparencia

```python
#Agrupar dados e verificar a quantidade
agrupado = dataset.groupby(['Aparencia']).size()
agrupado
```

```python
#Alterar "menos" por "sol" e verificar os dados da coluna
dataset.loc[dataset['Aparencia'] == "menos", 'Aparencia'] = "sol"
agrupado = dataset.groupby(['Aparencia']).size()
agrupado
```

Verificar coluna Temperatura

```python
#Verificar dados
dataset['Temperatura'].describe()
```

```python
#Verificar outliers
dataset.loc[(dataset['Temperatura'] < -130) | (dataset['Temperatura'] > 130)]
```

```python
#Calcular mediana
mediana = sts.median(dataset['Temperatura'])
mediana
```

```python
#Alterar os outliers pela mediana
dataset['Temperatura'] = dataset['Temperatura'].astype(float)
dataset.loc[(dataset['Temperatura'] < -130) | (dataset['Temperatura'] > 130), 'Temperatura'] = mediana
```

```python
#Verificar dados
dataset['Temperatura'].describe()
```

Verificar coluna Umidade

```python
#Verificar dados
dataset['Umidade'].describe()
```

```python
#Verificar outliers
dataset.loc[(dataset['Umidade'] < 0) | (dataset['Umidade'] > 100)]
```

```python
#Verificar a existencia de NAs
dataset['Umidade'].isnull().sum()
```

```python
#Calcular mediana
mediana2 = sts.median(dataset['Umidade'])
mediana2
```

```python
#Substituir outliers e NAs pela mediana
dataset.loc[(dataset['Umidade'] < 0) | (dataset['Umidade'] > 100), 'Umidade'] = mediana2
dataset['Umidade'] = dataset['Umidade'].fillna(mediana2)
```

```python
#Verificar dados
dataset['Umidade'].describe()
```

```python
#Verificar a existencia de NAs
dataset['Umidade'].isnull().sum()
```

Verificar coluna Vento

```python
#Agrupar dados
agrupado = dataset.groupby(['Vento']).size()
agrupado
```

```python
#Verificar a existencia de NAs
dataset['Vento'].isnull().sum()
```

```python
#Preencher o NAs com a moda (Falso)
dataset['Vento'] = dataset['Vento'].fillna("FALSO")
```

```python
#Verificar a existencia de NAs
dataset['Vento'].isnull().sum()
```

Verificar coluna Jogar

```python
#Agrupar dados para validar
agrupado = dataset.groupby(['Jogar']).size()
agrupado
```

```python
#Veriricar existencia de NAs
dataset['Jogar'].isnull().sum()
```
