```python
import pandas as pd
import seaborn as srn
import statistics as sts
```

Preparação dos Dados

```python
#Importar dados
dataset = pd.read_csv("data/Churn.csv", sep=";")
#Visualizar
dataset.head()
```

```python
#Visualizar tamanho
dataset.shape
```

```python
#Nomer as colunas
dataset.columns = ["Id", "Score", "Estado", "Genero", "Idade", "Parametro", "Saldo", "Produtos", "TemCartaoCredito", "Ativo", "Salario", "Saiu"] 
```

```python
dataset.head()
```

Analise Exploratória

Explorar Dados Categoricos

```python
#Estado
#Agrupando os dados da tabela Estado e verificando o tamanho
agrupado = dataset.groupby(['Estado']).size()
agrupado
```

```python
#Criando um grafico de barras
agrupado.plot.bar(color = 'gray')
```

```python
#Genero
#Agrupando os dados da tabela Genero e verificando o tamanho
agrupado = dataset.groupby(['Genero']).size()
agrupado
```

```python
#Criando um grafico de barras
agrupado.plot.bar(color = 'gray')
```
